```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load labeled data
df = pd.read_csv("dataset/labeled/aquarium_tinyml_dataset_labeled.csv")

# Split train/val/test
train, temp = train_test_split(df, test_size=0.3, stratify=df['label'], random_state=42)
val, test = train_test_split(temp, test_size=0.5, stratify=temp['label'], random_state=42)

train.to_csv("dataset/splits/train.csv", index=False)
val.to_csv("dataset/splits/val.csv", index=False)
test.to_csv("dataset/splits/test.csv", index=False)

print(f"Train: {len(train)}, Val: {len(val)}, Test: {len(test)}")
```
